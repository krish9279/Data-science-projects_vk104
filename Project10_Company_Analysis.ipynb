```python
import pymupdf
import cv2
import re
import pandas as pd
import fitz
import pymupdf
import cv2
import re
import pandas as pd
import numpy as np

# PDF path
pdf_path = r'C:\Users\kunni\Data-Science-Projects_v4\Astra\Pharma Magazines\National\EP_JANUARY_2026.pdf'

# Advertisement-related phrases
ad_keywords = [
    'advertisement', 'sponsored',
    'buy now', 'shop now', 'order now',
    'special offer', 'limited offer',
    'subscribe now', 'contact us',
    'scan here', 'scan to know more',
    'visit our website', 'book now',
    'register now', 'free trial',
    'discount', 'sale', 'promotional offer',
    'learn more', 'call today'
]


# 1. QR code detection
def detect_qr_code(page):

    pix = page.get_pixmap(
        matrix=pymupdf.Matrix(1.5, 1.5),
        alpha=False
    )

    image = np.frombuffer(
        pix.samples, dtype=np.uint8
    ).reshape(pix.height, pix.width, pix.n)

    image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)

    detector = cv2.QRCodeDetector()

    retval, decoded_info, points, _ = (
        detector.detectAndDecodeMulti(image)
    )

    return bool(retval and points is not None)


# 2. Advertisement detection
def detect_advertisement(page):

    text = page.get_text() or ''
    lower_text = text.lower()

    words = len(text.split())

    # Count website URLs and domain names
    urls = len(re.findall(
        r'https?://|www\.|[\w-]+\.(?:com|in|org|net|co\.in)\b',
        lower_text
    ))

    # Count promotional keywords
    keyword_count = sum(
        lower_text.count(keyword)
        for keyword in ad_keywords
    )

    # Estimate image coverage
    page_area = page.rect.width * page.rect.height

    image_blocks = page.get_image_info()

    image_area = sum(
        max(0, info['bbox'][2] - info['bbox'][0]) *
        max(0, info['bbox'][3] - info['bbox'][1])
        for info in image_blocks
    )

    image_ratio = (
        min(image_area / page_area, 1)
        if page_area else 0
    )

    # QR code detection
    has_qr = detect_qr_code(page)

    # Advertisement score
    score = 0

    if words < 35:
        score += 2
    elif words < 80:
        score += 1

    if urls >= 2:
        score += 2
    elif urls == 1:
        score += 1

    if keyword_count >= 2:
        score += 2
    elif keyword_count == 1:
        score += 1

    if image_ratio > 0.35:
        score += 1

    if has_qr:
        score += 3

    is_ad = score >= 4

    return {
        'content': text,
        'word_count': words,
        'URLs': urls,
        'promotional_keywords': keyword_count,
        'image_ratio': round(image_ratio, 2),
        'QR_code': has_qr,
        'ad_score': score,
        'is_ad': is_ad
    }


# 3. Process PDF
doc = pymupdf.open(pdf_path)

results = []

start_page = 52
end_page = min(376, len(doc))

if start_page >= len(doc):
    raise ValueError("Starting page exceeds PDF length")

for i in range(start_page, end_page):

    page = doc[i]

    result = detect_advertisement(page)

    result['Page'] = i + 1

    results.append(result)


# 4. Create DataFrame
df_results = pd.DataFrame(results)


# 5. Separate advertisements and articles
advertisements = df_results[
    df_results['is_ad'] == True
].copy()

articles = df_results[
    df_results['is_ad'] == False
].copy()


# 6. Retain article content
df_i = articles[['Page', 'content']].copy()

# Remove unnecessary newlines and spaces
df_i['content'] = df_i['content'].apply(
    lambda x: ' '.join(x.split())
)


# 7. Display results
print("Total pages processed:", len(results))
print("Advertisement pages:", len(advertisements))
print("Article pages:", len(articles))

display(df_i.head())
```

```python
nlp = spacy.load("en_core_web_md")

keywords = [nlp(word) for word in ["Astrazeneca","Oncology", "Cardiovascular", "Renal","Rare Disease","Infectious Disease","BioPharmaceuticals"]]

def max_similarity(sentence):
    doc = nlp(str(sentence))
    return max(keyword.similarity(doc) for keyword in keywords)
```

```python
def review_rating(string):
    scores = sent.polarity_scores(str(string))
    if scores['compound'] == 0:
        return 'Neutral'
    elif scores['compound'] > 0:
        return 'Positive'
    else:
        return 'Negative'

dfs['Sentiment'] = dfs['PharmaTimes-Text'].apply(review_rating)
```

```python

# ============================================================
# PRODUCT IMAGE CLASSIFICATION USING EFFICIENTNETB0
# TRANSFER LEARNING + FINE-TUNING
# ============================================================

import os
import re
import json
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

from tensorflow.keras import layers, models


# ============================================================
# 1. CONFIGURATION
# ============================================================

# Change this path according to your dataset location

DATASET_PATH = r'C:\Users\kunni\Data-Science-Projects_v4\Astra\Drug Vision\Data Combined\All'

IMG_SIZE = 224
BATCH_SIZE = 32

INITIAL_EPOCHS = 20
FINE_TUNE_EPOCHS = 10

RANDOM_STATE = 42

MODEL_PATH = 'product_classifier.keras'
CLASS_NAMES_PATH = 'product_classes.json'


# Check TensorFlow and GPU

print("TensorFlow version:", tf.__version__)

print("Available GPUs:",
      tf.config.list_physical_devices('GPU'))


# ============================================================
# 2. LOAD IMAGE PATHS AND EXTRACT PRODUCT LABELS
# ============================================================

data = []

for filename in os.listdir(DATASET_PATH):

    if filename.lower().endswith(
        ('.jpg', '.jpeg', '.png')
    ):

        image_path = os.path.join(
            DATASET_PATH,
            filename
        )

      

        product_name = os.path.splitext(filename)[0]

        

        product_name = re.sub(
            r'\s*\(\d+\)$',
            '',
            product_name
        )

        product_name = product_name.strip().upper()

        data.append([
            image_path,
            product_name
        ])


df = pd.DataFrame(
    data,
    columns=['image_path', 'product']
)


print("\nTotal images:", len(df))

print("Total product classes:",
      df['product'].nunique())


print("\nProduct distribution:")

print(df['product'].value_counts())


# Check dataset

if len(df) == 0:

    raise ValueError("No images found in dataset folder.")

if df['product'].nunique() < 2:

    raise ValueError("At least two product classes are required.")


# ============================================================
# 3. CHECK NUMBER OF IMAGES PER PRODUCT
# ============================================================

counts = df['product'].value_counts()

print("\nProducts with fewer than 7 images:")

print(counts[counts < 7])

if counts.min() < 7:

    raise ValueError(
        "Some product classes have fewer than 7 images. "
        "Add more images or use a different split strategy."
    )


# ============================================================
# 4. SPLIT DATASET
# TRAINING 70%, VALIDATION 15%, TESTING 15%
# ============================================================

train_val_df, test_df = train_test_split(

    df,

    test_size=0.15,

    random_state=RANDOM_STATE,

    stratify=df['product']

)


train_df, val_df = train_test_split(

    train_val_df,

    test_size=0.17647,

    random_state=RANDOM_STATE,

    stratify=train_val_df['product']

)


print("\nDataset split:")

print("Training images:", len(train_df))

print("Validation images:", len(val_df))

print("Testing images:", len(test_df))


# ============================================================
# 5. ENCODE PRODUCT LABELS
# ============================================================

class_names = sorted(
    df['product'].unique()
)

class_to_id = {

    name: index

    for index, name in enumerate(class_names)

}


num_classes = len(class_names)


for dataframe in [train_df, val_df, test_df]:

    dataframe['label'] = dataframe['product'].map(
        class_to_id
    )


print("\nProduct classes:")

print(class_names)

print("\nNumber of classes:", num_classes)


# ============================================================
# 6. IMAGE LOADING AND PREPROCESSING
# ============================================================

def load_image(path, label):

    image = tf.io.read_file(path)

    image = tf.io.decode_image(

        image,

        channels=3,

        expand_animations=False

    )

    image.set_shape([None, None, 3])

    # Resize while preserving aspect ratio

    image = tf.image.resize_with_pad(

        image,

        IMG_SIZE,

        IMG_SIZE

    )

    image = tf.cast(
        image,
        tf.float32
    )

    return image, label


AUTOTUNE = tf.data.AUTOTUNE


# ============================================================
# 7. CREATE TENSORFLOW DATASETS
# ============================================================

def create_dataset(dataframe, training=False):

    paths = dataframe['image_path'].values

    labels = dataframe['label'].values.astype(
        np.int32
    )

    dataset = tf.data.Dataset.from_tensor_slices(

        (paths, labels)

    )

    dataset = dataset.map(

        load_image,

        num_parallel_calls=AUTOTUNE

    )

    if training:

        dataset = dataset.shuffle(

            buffer_size=len(dataframe),

            seed=RANDOM_STATE

        )

    dataset = dataset.batch(BATCH_SIZE)

    dataset = dataset.prefetch(AUTOTUNE)

    return dataset


train_ds = create_dataset(
    train_df,
    training=True
)

val_ds = create_dataset(
    val_df
)

test_ds = create_dataset(
    test_df
)


# ============================================================
# 8. DATA AUGMENTATION
# ============================================================

# Mild augmentation to handle different photographs,
# backgrounds, lighting and viewing angles.
#
# Avoid horizontal/vertical flips because package text
# and product orientation can be important.

data_augmentation = tf.keras.Sequential([

    layers.RandomRotation(0.03),

    layers.RandomZoom(0.10),

    layers.RandomTranslation(
        height_factor=0.05,
        width_factor=0.05
    ),

    layers.RandomContrast(0.10)

])


# ============================================================
# 9. LOAD PRETRAINED EFFICIENTNETB0
# ============================================================

base_model = tf.keras.applications.EfficientNetB0(

    weights='imagenet',

    include_top=False,

    input_shape=(IMG_SIZE, IMG_SIZE, 3)

)


# Freeze pretrained layers initially

base_model.trainable = False


# ============================================================
# 10. BUILD CLASSIFICATION MODEL
# ============================================================

inputs = layers.Input(

    shape=(IMG_SIZE, IMG_SIZE, 3)

)


# Apply data augmentation

x = data_augmentation(inputs)


# Extract deep visual features

x = base_model(

    x,

    training=False

)


# Convert feature maps into feature vectors

x = layers.GlobalAveragePooling2D()(x)


# Batch normalization

x = layers.BatchNormalization()(x)


# Dropout to reduce overfitting

x = layers.Dropout(0.35)(x)


# Product classification output

outputs = layers.Dense(

    num_classes,

    activation='softmax'

)(x)


model = models.Model(

    inputs=inputs,

    outputs=outputs

)


# ============================================================
# 11. COMPILE MODEL
# ============================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(

        learning_rate=0.001

    ),

    loss='sparse_categorical_crossentropy',

    metrics=['accuracy']

)


print("\nMODEL SUMMARY")

model.summary()


# ============================================================
# 12. CALLBACKS
# ============================================================

def get_callbacks(checkpoint_path):

    return [

        tf.keras.callbacks.EarlyStopping(

            monitor='val_loss',

            patience=5,

            restore_best_weights=True

        ),

        tf.keras.callbacks.ModelCheckpoint(

            checkpoint_path,

            monitor='val_accuracy',

            save_best_only=True

        ),

        tf.keras.callbacks.ReduceLROnPlateau(

            monitor='val_loss',

            factor=0.2,

            patience=2,

            min_lr=1e-7

        )

    ]


# ============================================================
# 13. INITIAL TRAINING
# FEATURE EXTRACTION
# ============================================================

print("\nINITIAL TRAINING STARTED")

history = model.fit(

    train_ds,

    validation_data=val_ds,

    epochs=INITIAL_EPOCHS,

    callbacks=get_callbacks(
        'best_feature_extractor.keras'
    )

)


# ============================================================
# 14. FINE-TUNING
# UNFREEZE LAST 30 LAYERS
# ============================================================

print("\nFINE-TUNING STARTED")


base_model.trainable = True


# Freeze earlier layers

for layer in base_model.layers[:-30]:

    layer.trainable = False


# Keep Batch Normalization layers frozen

for layer in base_model.layers:

    if isinstance(
        layer,
        tf.keras.layers.BatchNormalization
    ):

        layer.trainable = False


# Recompile with a much smaller learning rate

model.compile(

    optimizer=tf.keras.optimizers.Adam(

        learning_rate=0.00001

    ),

    loss='sparse_categorical_crossentropy',

    metrics=['accuracy']

)


history_fine = model.fit(

    train_ds,

    validation_data=val_ds,

    epochs=FINE_TUNE_EPOCHS,

    callbacks=get_callbacks(
        'best_finetuned_model.keras'
    )

)


# ============================================================
# 15. LOAD BEST FINE-TUNED MODEL
# ============================================================

model = tf.keras.models.load_model(

    'best_finetuned_model.keras'

)


# ============================================================
# 16. PLOT TRAINING AND VALIDATION ACCURACY
# ============================================================

train_accuracy = (
    history.history['accuracy']
    + history_fine.history['accuracy']
)

val_accuracy = (
    history.history['val_accuracy']
    + history_fine.history['val_accuracy']
)


plt.figure(figsize=(10, 5))

plt.plot(
    train_accuracy,
    label='Training Accuracy'
)

plt.plot(
    val_accuracy,
    label='Validation Accuracy'
)

plt.xlabel('Epoch')

plt.ylabel('Accuracy')

plt.title('EfficientNetB0 Training Accuracy')

plt.legend()

plt.grid()

plt.show()


# ============================================================
# 17. PLOT TRAINING AND VALIDATION LOSS
# ============================================================

train_loss = (
    history.history['loss']
    + history_fine.history['loss']
)

val_loss = (
    history.history['val_loss']
    + history_fine.history['val_loss']
)


plt.figure(figsize=(10, 5))

plt.plot(
    train_loss,
    label='Training Loss'
)

plt.plot(
    val_loss,
    label='Validation Loss'
)

plt.xlabel('Epoch')

plt.ylabel('Loss')

plt.title('EfficientNetB0 Training Loss')

plt.legend()

plt.grid()

plt.show()


# ============================================================
# 18. EVALUATE MODEL ON TEST DATA
# ============================================================

print("\nMODEL EVALUATION")


y_true = np.concatenate([

    labels.numpy()

    for _, labels in test_ds

])


y_prob = model.predict(

    test_ds,

    verbose=1

)


y_pred = np.argmax(

    y_prob,

    axis=1

)


accuracy = accuracy_score(

    y_true,

    y_pred

)


print(

    "\nTEST ACCURACY:",

    round(accuracy * 100, 2),

    "%"

)


# ============================================================
# 19. CLASSIFICATION REPORT
# ============================================================

print("\nCLASSIFICATION REPORT\n")


print(

    classification_report(

        y_true,

        y_pred,

        labels=np.arange(num_classes),

        target_names=class_names,

        zero_division=0

    )

)


# ============================================================
# 20. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_true,

    y_pred,

    labels=np.arange(num_classes)

)


plt.figure(figsize=(14, 10))


sns.heatmap(

    cm,

    annot=True,

    fmt='d',

    cmap='Blues',

    xticklabels=class_names,

    yticklabels=class_names

)


plt.xlabel('Predicted Product')

plt.ylabel('Actual Product')

plt.title('Product Classification Confusion Matrix')

plt.xticks(rotation=90)

plt.yticks(rotation=0)

plt.tight_layout()

plt.show()


# ============================================================
# 21. SAVE MODEL
# ============================================================

model.save(MODEL_PATH)


with open(CLASS_NAMES_PATH, 'w') as f:

    json.dump(class_names, f)


print("\nModel saved:", MODEL_PATH)

print("Class names saved:", CLASS_NAMES_PATH)


# ============================================================
# 22. PREDICT A NEW PRODUCT IMAGE
# ============================================================

def predict_product(image_path):

    # Read image

    image = tf.io.read_file(image_path)

    image = tf.io.decode_image(

        image,

        channels=3,

        expand_animations=False

    )

    image.set_shape([None, None, 3])


    # Preserve original image for display

    original_image = image.numpy()


    # Apply the same preprocessing as training

    processed_image = tf.image.resize_with_pad(

        image,

        IMG_SIZE,

        IMG_SIZE

    )


    processed_image = tf.cast(

        processed_image,

        tf.float32

    )


    processed_image = tf.expand_dims(

        processed_image,

        axis=0

    )


    # Predict probabilities

    probabilities = model.predict(

        processed_image,

        verbose=0

    )[0]


    # Find top 3 predictions

    top_indices = np.argsort(

        probabilities

    )[::-1][:3]


    # Display image

    plt.figure(figsize=(6, 6))

    plt.imshow(original_image)

    plt.axis('off')

    plt.title('Test Product Image')

    plt.show()


    print("\nPRODUCT PREDICTIONS\n")


    for idx in top_indices:

        product_name = class_names[idx]

        confidence = probabilities[idx] * 100

        print(

            f"{product_name}: "

            f"{confidence:.2f}%"

        )


    predicted_product = class_names[

        top_indices[0]

    ]


    print("\nPREDICTED PRODUCT:")

    print(predicted_product)


    return predicted_product


# ============================================================
# 23. TEST WITH A NEW IMAGE
# ============================================================

test_image_path = r'C:\Users\kunni\Data-Science-Projects_v4\Astra\Drug Vision\Data Combined\alaxan.jpg'


predicted_product = predict_product(

    test_image_path

)


# ============================================================
# END OF PROGRAM
# ============================================================
```
